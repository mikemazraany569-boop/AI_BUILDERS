```python
import torch.nn as nn
```

```python
import torch.nn.functional as f
```

```python
class Model(nn.Module):
    def __init__(self,in_features=4,h1=8,h2=9,out_features=3):
        super().__init__()
        self.fc1=nn.Linear(in_features,h1)
        self.fc2=nn.Linear(h1,h2)
        self.out=nn.Linear(h2,out_features)
        
    def forward(self,x):
        x=f.relu(self.fc1(x))
        x=f.relu(self.fc2(x))
        x=self.out(x)
        return x
```

```python
model=Model()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
url = "https://gist.githubusercontent.com/curran/a08a1080b88344b0c8a7/raw/iris.csv"

my_df = pd.read_csv(url)

my_df.head()
```

```python
my_df.tail()
```

```python
my_df['species']=my_df['species'].replace('setosa',0)
my_df['species']=my_df['species'].replace('versicolor',1)
my_df['species']=my_df['species'].replace('virginica',2)
my_df
```

```python
X=my_df.drop('species',axis=1)
y=my_df['species']
```

```python
X=X.values
y=y.values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=41)
```

```python
X_train=torch.FloatTensor(X_train)
X_test=torch.FloatTensor(X_test)
```

```python
y_train = y_train.astype(int)
y_test = y_test.astype(int)
```

```python
y_train=torch.LongTensor(y_train)
y_test=torch.LongTensor(y_test)
```

```python
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=0.01)
```

```python
model.parameters
```

```python
epochs=100
losses=[]
for i in range(epochs):
    y_pred=model.forward(X_train)
    loss=criterion(y_pred,y_train)
    losses.append(loss.detach().numpy())
    if i%10==0:
        print(f'Epoch:{i} and loss:{loss}')
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()    
```

```python
plt.plot(range(epochs),losses)
plt.ylabel('loss/error')
plt.xlabel('Epoch')
```

```python
with torch.no_grad():
    y_eval=model.forward(X_test)
    loss=criterion(y_eval,y_test)
```

```python
loss
```

```python
correct=0
with torch.no_grad():
    for i,data in enumerate(X_test):
        y_val=model.forward(data)
        if y_test[i]==0:
            x='Setosa'
        elif y_test[i]==1:
            x='Versicolor'
        else:
            x='Virginica'
        print(f'{i+1}.str{(y_val)}\t {y_test[i]}\t {y_val.argmax().item()}\t {x}')
        if y_val.argmax().item()==y_test[i]:
            correct+=1
print(f'We got {correct} correct')            
```
